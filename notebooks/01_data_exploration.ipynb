```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/raw/equipment_sensor_data.csv")

df.head()

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df["failure"].value_counts().plot(kind="bar")

plt.title("Normal vs Failure")
plt.xlabel("Failure")
plt.ylabel("Number of Machines")
plt.show()
```

```python
df.hist(figsize=(12, 10))

plt.tight_layout()
plt.show()
```

```python
#check the correlation 
plt.figure(figsize=(10, 7))

sns.heatmap(df.corr(), annot=True, cmap="coolwarm")

plt.title("Feature Correlation")
plt.show()
```

```python
# compare vibration

sns.boxplot(data=df, x="failure", y="vibration")

plt.title("Vibration vs Failure")
plt.xlabel("Failure")
plt.ylabel("Vibration")
plt.show()
```

```python
# compare temp

sns.boxplot(data=df, x="failure", y="engine_temperature")

plt.title("Engine Temperature vs Failure")
plt.xlabel("Failure")
plt.ylabel("Engine Temperature")
plt.show()
```
