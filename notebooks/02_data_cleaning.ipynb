```python
import pandas as pd
#load the dataset
df = pd.read_csv("../data/raw/equipment_sensor_data.csv")

df.head()
```

```python
# check for missing values
df.isnull().sum()
```

```python
#remove the duplicates
print("Duplicates:", df.duplicated().sum())

df = df.drop_duplicates()

print("Rows after cleaning:", len(df))
```

```python
# check the datatypes
df.dtypes
```

```python
# check for impossible values ie they shouldnt be negative

sensor_columns = [
    "engine_temperature",
    "oil_temperature",
    "hydraulic_pressure",
    "vibration",
    "engine_rpm",
    "fuel_consumption",
    "operating_hours"
]

(df[sensor_columns] < 0).sum()
```

```python
# seperate features and target
X = df.drop(columns=["failure"])
y = df["failure"]

print("Features:")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)
```

```python
#check the target
print(y.value_counts())
print()
print(y.value_counts(normalize=True))
```

```python
from pathlib import Path
output_path = Path("../data/processed/clean-equipment_data.csv")
output_path.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(output_path, index=False)

print(f"Saved cleaned dataset to: {output_path}")
```
